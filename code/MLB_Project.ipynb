```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

data = pd.read_csv('MLB_Data(Worksheet).csv')
data_velo = pd.read_csv('savant_data.csv')
```

```python
print(data)
```

```python
data_velo = data_velo.rename(columns = {'year':'Year'})
print(data_velo)
```

```python
bar = sns.catplot(data=data_velo, aspect=3, x="Year",y="velocity",kind="bar" , color="blue")
bar.set(ylim=(91, 95))
bar.fig.suptitle("MLB Batting Average Over Time")
```

```python
plt.figure()
plt.plot(data["Year"], data["BA"], linewidth=2)
plt.ylabel("Batting Average")
plt.title("MLB Batting Average Over Time")
plt.grid(True)
plt.show
```

```python
plt.figure()
plt.plot(data["Year"], data["SO"], linewidth=2)
plt.ylabel("Strike Outs")
plt.title("MLB Stikeouts Per Game")
plt.grid(True)
plt.show
```

```python
correlation = data[["BA", "SO","ERA","TB"]].corr()
plt.figure()
sns.heatmap(correlation, annot=True, cmap="coolwarm",)
plt.title("Correlation Analysis")
plt.show
```

```python
merged = pd.merge(data_velo, data, on="Year", how="inner")

plt.figure()
plt.scatter(merged['velocity'], merged['BA'])
plt.xlabel("Average Pitch Velocity(MPH)")
plt.ylabel("Batting Average")
plt.title("Batting Average VS Pitch Velocity(2008-2025)")
plt.grid(True)
plt.show()
```
